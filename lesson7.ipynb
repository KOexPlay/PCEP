```python
'''Write a program to Match key values in two dictionaries
Sample Output
A = {‘Spanish’ : 92, 'English' : 56, 'Maths' : 88, 'Science' : 97, 'Social' : 89}
B= {'Spanish' : 78, 'English' : 68, 'Maths' : 88, 'Sceince' : 97, 'Social' : 56}
Maths : 88 is present in both A and B'''
```

```python
A = {'Spanish' : 92, 'English' : 56, 'Maths' : 88, 'Science' : 97, 'Social' : 89}
B= {'Spanish' : 78, 'English' : 68, 'Maths' : 88, 'Science' : 97, 'Social' : 56}

print(A)
print(B)
for lang in A:
    if A[lang] == B[lang]:
        print(f"{lang} : {A[lang]} is present in both A and B")
```
